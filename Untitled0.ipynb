```python
import numpy as np

# 1. Create a numpy array of size 2 x 2
arr1 = np.array([[1, 2], [3, 4]])
print("--- 1. 2x2 Array ---")
print(arr1)
print()

# 2. Create a random numpy array of size 3 x 4
arr2 = np.random.rand(3, 4)
print("--- 2. Random 3x4 Array ---")
print(arr2)
print()

# 3. Create a numpy array of ones and a numpy array of zeroes
ones_arr = np.ones((2, 3))
zeros_arr = np.zeros((2, 3))
print("--- 3. Ones Array ---")
print(ones_arr)
print("--- 3. Zeros Array ---")
print(zeros_arr)
print()

# 4. Create a random numpy array of size 6x3 and transpose it
arr4 = np.random.rand(6, 3)
arr4_transposed = arr4.T
print("--- 4. Original 6x3 Array ---")
print(arr4)
print("--- 4. Transposed 3x6 Array ---")
print(arr4_transposed)
print()

# 5. Create two random numpy array of size n x p and p x m and then compute the dot product
n, p, m = 3, 4, 2
A = np.random.rand(n, p)
B = np.random.rand(p, m)
dot_product = np.dot(A, B)

print("--- 5. Matrix A (n x p) ---")
print(A)
print("--- 5. Matrix B (p x m) ---")
print(B)
print("--- 5. Dot Product (n x m) ---")
print(dot_product)
```
